## Subspace-Targeted Distillation for Domain-Specific Students

### Motivation

**General-purpose LLMs** represent vast amounts of knowledge distributed across high-dimensional parameter spaces. When distilling a small model for a specialized domain (e.g., Code, Medical, or Finance), forcing the student to learn the teacher's full parameter distribution wastes capacity.

**Subspace-Targeted Distillation** restricts the distillation process to the specific activation subspaces and representation manifolds activated by domain-specific prompts.

## Core Components & Mathematical Formulation

**Subspace-targeted distillation** aligns internal representations in addition to output logits.

### A. Hidden-State Projection Loss ($\mathcal{L}_{\text{Subspace}}$)

Because the Student and Teacher have different hidden dimensions ($d_S < d_T$), we introduce a trainable linear transformation matrix $W_{\text{proj}} \in \mathbb{R}^{d_S \times d_T}$ to map student states into the teacher's domain subspace:

$$\mathcal{L}_{\text{Hidden}} = \frac{1}{L} \sum_{l \in \mathcal{S}} \left\Vert{} h_T^{(l)} - W_{\text{proj}} \, h_S^{(l)} \right\Vert{}_2^2$$

**Where** $\mathcal{S}$ represents key domain-sensitive transformer layers (e.g., mid-to-late attention layers).
